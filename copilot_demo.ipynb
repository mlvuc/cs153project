# Energy Copilot — Conversational Demo

Shows the Claude-powered copilot layer on top of the scheduler engine.
Submit jobs in plain English; the copilot calls the scheduler and explains the recommendation.

> **Requires** `ANTHROPIC_API_KEY` in your `.env` file.

```python
from copilot.planner import EnergyCopilot

copilot = EnergyCopilot()
```

## Job 1 — ETL pipeline

```python
response = copilot.chat(
    "I need to run a 2-hour ETL pipeline. It has to finish by 2pm today. "
    "When should I start it to get the cheapest energy?"
)
print(response)
```

## Job 2 — Model training (multi-turn)

```python
copilot.reset()  # fresh conversation

response = copilot.chat(
    "We have a 6-hour model training run that needs to complete before 9am tomorrow. "
    "What's the optimal window?"
)
print(response)
```

```python
# Follow-up in the same conversation
response = copilot.chat("What if we don't care about cost and just want the lowest carbon window instead?")
print(response)
```

## Job 3 — Urgent data export

```python
copilot.reset()

response = copilot.chat(
    "Quick question — I have a 1-hour data export that has to be done in 3 hours. "
    "Is there any benefit to waiting, or should I just kick it off now?"
)
print(response)
```
